```python
from qiskit.quantum_info import SparsePauliOp
from sympy import I, Matrix, latex
from IPython.display import display, Math
import numpy as np
```

```python
H = SparsePauliOp(data=["IIII", 
                        "ZIII", "IZII", "IIZI", 
                        "ZZII", "ZIZI", "IZIZ", 
                        "XZXI", "YZYI", 
                        "ZZZI", "ZIZZ", "IZZZ", 
                        "XZXZ", "YZYZ", 
                        "ZZZZ"],
                  coeffs=[-0.81261, 
                          0.171201, 0.16862325, -0.2227965, 
                          0.171201, 0.12054625, 0.17434925, 
                          0.04532175, 0.04532175,
                          0.165868, 0.12054625, -0.2227965, 
                          0.04532175, 0.04532175,
                        0.165868])
display(Math(latex(Matrix(H))))
eigvals, eigenvects = np.linalg.eig(H.to_matrix())
sorted(eigvals)
```
